1.1- Consulta la API pública JSONPlaceholder utilizando el método GET.

```python
url_posts = "https://jsonplaceholder.typicode.com/posts"        # Se guarda en una variable para si cambia la URL solo modificar 1 línea
respuesta_posts = requests.get(url_posts)                 # Petición HTTP
posts = respuesta_posts.json()                   # El método .json() convierte ese JSON en objetos de Python.
posts[0]
```

```python
url_users = "https://jsonplaceholder.typicode.com/users"

respuesta_users = requests.get(url_users)

users = respuesta_users.json()

users[0]
```

```python
url_todos = "https://jsonplaceholder.typicode.com/todos"

respuesta_todos = requests.get(url_todos)

todos = respuesta_todos.json()

todos[0]
```

1.2- Muestra por pantalla:

```python
# Cantidad de recursos
print("Número de publicaciones:", len(posts))     # posts, users y todos son listas, podemos utilizar la función len() para saber cuántos elementos contienen.
print("Número de usuarios:", len(users))
print("Número de tareas:", len(todos))
```

```python
# Código de estado
print("Código de estado de publicaciones:", respuesta_posts.status_code)
print("Código de estado de usuarios:", respuesta_users.status_code)
print("Código de estado de tareas:", respuesta_todos.status_code)
```

1.3- 3. Realiza una petición a una publicación inexistente para obtener un error 404 y muestra el código de estado recibido.

```python
url_post_inexistente = "https://jsonplaceholder.typicode.com/posts/991"

respuesta_404 = requests.get(url_post_inexistente)

print("Código de estado:", respuesta_404.status_code)   # Porque el servidor recibe una petición válida (GET), pero el recurso solicitado (/posts/991) no existe
```

1.4- Realiza una petición POST para crear una nueva publicación ficticia. Incluye el título, el cuerpo del mensaje y un userId.
Muestra la respuesta JSON.
Muestra el código de estado.

```python
url_posts = "https://jsonplaceholder.typicode.com/posts"

nueva_publicacion = {                                            # Los datos se crean como un diccionario de Python.
    "title": "Mi primera publicación",
    "body": "Esta publicación ha sido creada desde Python utilizando requests.",
    "userId": 1
}

respuesta_post = requests.post(url_posts, json=nueva_publicacion)    # indica a qué recurso enviamos la petición y convierte el diccionario a JSON y lo envía en el cuerpo (body) de la petición.

print("Respuesta JSON:")
print(respuesta_post.json())

print()

print("Código de estado:", respuesta_post.status_code)    # Devuelve 201 Created (EL recurso se ha creado correctamente)
```

1.5- 5. Realiza una petición PATCH para modificar parcialmente una publicación existente.
Muestra la respuesta JSON.
Muestra el código de estado.

```python
url_patch = "https://jsonplaceholder.typicode.com/posts/1"    # Se modifica la publicación con id = 1.

datos_actualizados = {
    "title": "Título modificado desde Python"
}

respuesta_patch = requests.patch(url_patch, json=datos_actualizados)

print("Respuesta JSON:")
print(respuesta_patch.json())

print()

print(f"Código de estado: {respuesta_patch.status_code}")   # El código 200 OK significa que el servidor ha procesado la petición con éxito y devuelve el recurso actualizado
```

1.6- Haz una petición DELETE sobre una publicación.
Muestra la respuesta JSON.
Muestra el código de estado.

```python
url_delete = "https://jsonplaceholder.typicode.com/posts/1"

respuesta_delete = requests.delete(url_delete)        # No es necesario enviar un body, porque el recurso que queremos eliminar ya está identificado en la URL.

print("Respuesta JSON:")
print(respuesta_delete.json())

print(f"Código de estado: {respuesta_delete.status_code}")    # Devuelve 200 OK (la eliminación fue correcta y el servidor devuelve alguna información)
```

**2.1** - Interacción con una API pública real

Para esta práctica se ha seleccionado la API **Open-Meteo**, disponible en el repositorio **Public APIs** de GitHub.

Esta API proporciona información meteorológica mediante peticiones HTTP GET, devuelve los datos en formato JSON y no requiere autenticación mediante API Key, por lo que resulta adecuada para trabajar con la librería `requests` de Python.

***2.2.1***- ## Endpoints disponibles

Tras revisar la documentación de Open-Meteo, se han identificado los siguientes endpoints:

- **Weather Forecast**: permite obtener la previsión meteorológica para una ubicación determinada.
- **Air Quality**: proporciona información sobre la calidad del aire y diferentes contaminantes atmosféricos.

***2.2.2***- Parámetros de la API

La API Open-Meteo permite utilizar diferentes parámetros para personalizar las consultas. Algunos de los más relevantes son:

- latitude: indica la latitud de la ubicación de la consulta.
- longitude: indica la longitud de la ubicación.
- current: permite solicitar datos meteorológicos actuales, como la temperatura o la velocidad del viento.
- hourly: permite obtener variables meteorológicas con frecuencia horaria.
- daily: permite obtener información meteorológica diaria, como la temperatura máxima y mínima.
- timezone: permite especificar la zona horaria en la que se desean recibir los datos.

***2.2.3***- Tras revisar la documentación de la API Open-Meteo, se ha comprobado que las respuestas se devuelven en formato **JSON (JavaScript Object Notation)**.

Este formato organiza la información en pares **clave-valor**, como por ejemplo 'latitude', 'longitude'', 'timezone' y 'hourly', lo que facilita el procesamiento de los datos desde Python utilizando la librería 'requests' y su posterior conversión a un DataFrame de pandas

**2.3**-  Haz una petición GET sencilla:
Muestra el código de estado de la respuesta.
Imprime de forma clara algunos de los campos de la respuesta JSON.

```python
# URL de la API
url = "https://api.open-meteo.com/v1/forecast"

# Parámetros de la consulta
parametros = {
    "latitude": 41.39,
    "longitude": 2.17,
    "current": "temperature_2m"
}

# Petición GET
respuesta = requests.get(url, params=parametros)

# Código de estado
print(f"Código de estado: {respuesta.status_code}")

# Convertir la respuesta JSON a un diccionario
datos = respuesta.json()

# Mostrar algunos campos de la respuesta
print(f"Latitud: {datos['latitude']}")
print(f"Longitud: {datos['longitude']}")
print(f"Zona horaria: {datos['timezone']}")
print(f"Temperatura actual: {datos['current']['temperature_2m']} °C")
```

***2.4***  Convierte los datos a un DataFrame de pandas:

Muestra las primeras filas de DataFrame.

```python
import pandas as pd

# Convertir los datos actuales en un DataFrame
df_current = pd.DataFrame([datos["current"]])

# Mostrar las primeras filas
df_current.head()
```

```python
df = pd.DataFrame([datos])
df.head()
```

**3.1** Buscar un dataset de tu interés mediante package_search y package_show.

```python
import requests

url = "https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_search"

parametros = {
    "q": "bicing"
}

respuesta = requests.get(url, params=parametros)

print(f"Código de estado: {respuesta.status_code}")

datos = respuesta.json()

# Mostrar el número de resultados encontrados
print(f"Número de datasets encontrados: {datos['result']['count']}")

# Mostrar el título de los primeros 5 datasets
for dataset in datos["result"]["results"][:5]:
    print(dataset["title"])
```

```python
for dataset in datos["result"]["results"]:         # package_show necesita el nombre interno (name), no el título.
    print(f"Título: {dataset['title']}")
    print(f"Nombre: {dataset['name']}")
    print("-" * 50)
```

```python
url_show = "https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_show"

parametros = {
    "id": "estat-estacions-bicing"
}

respuesta_show = requests.get(url_show, params=parametros)

print(f"Código de estado: {respuesta_show.status_code}")

datos_show = respuesta_show.json()
```

```python
print(f"Título: {datos_show['result']['title']}")        # Campos interesantes
print(f"Autor: {datos_show['result']['author']}")
print(f"Organización: {datos_show['result']['organization']['title']}")
print(f"Número de recursos: {len(datos_show['result']['resources'])}")
```

```python
for recurso in datos_show["result"]["resources"]:         # Muestra los archivos que tiene el dataset
    print(f"Nombre: {recurso['name']}")
    print(f"Formato: {recurso['format']}")
    print(f"URL: {recurso['url']}")
    print("-" * 50)
```

Se ha utilizado el endpoint package_search para localizar un dataset relacionado con el servicio Bicing de Barcelona. Posteriormente, mediante package_show, se ha recuperado la información detallada del dataset, incluyendo su autor, organización responsable y los recursos disponibles para su descarga en diferentes formatos.

**3.2** De los resultados obtenidos, selecciona uno que tenga recursos disponibles en CSV o JSON.

```python
# Seleccionar el primer recurso en formato JSON
recurso = datos_show["result"]["resources"][0]

print(f"Nombre: {recurso['name']}")
print(f"Formato: {recurso['format']}")
print(f"URL: {recurso['url']}")
```

1. Calculadora del índice de masa corporal

```python
def calcular_imc(peso,altura):
    ''' Calcula el IMC a partir de dos parámetros introducidos, 
    devuelve un mensaje con el índice y la clasificación según el mismo.'''
    
    resultado = (peso) / (altura**2)

    if resultado < 18.5 :
        return f'Su IMC es de {resultado:.2f} y muestra un peso Bajo!'
    elif resultado <=24.9:
        return f'Su IMC es de {resultado:.2f} y muestra un peso Normal!'
    elif resultado <=29.9:
        return f'Su IMC es de {resultado:.2f} y muestra un Sobrepeso!'
    else:
        return f'Su IMC es de {resultado:.2f} y muestra Obesidad!'
    
def pedir_validar(mensaje, minimo, maximo):
    '''Solicita un número y valida que sea numérico y que esté dentro del rango permitido.'''

    while True:

        try:
            numero = float(input(mensaje))

            if minimo <= numero <= maximo:
                return numero
            else:
                print(f'El valor debe estar entre {minimo} y {maximo}.')

        except ValueError:
            print('Debe introducir un número.')


peso = pedir_validar('Introduce su peso en kg: ', 10, 400)

altura = pedir_validar('Introduce su altura en m: ', 0.50, 2.50)

print(calcular_imc(peso, altura))

```

2. Conversor de temperaturas

```python
def conversor_temp(grados, unidad):
    ''' Convierte la temperatura a otras dos medidas
        a partir de dos parámetros solicitados y devuelve el resultado.'''
    conversiones ={}

    if unidad == 'c':
        resultado_f = (grados * 9/5) + 32 
        resultado_k = (grados + 273.15)
        conversiones['Fahrenheit'] = resultado_f       # ESCRIBO EL NOMBRE COMPLETO DE LA UNIDAD COMO CLAVE PARA QUE LUEGO SE MUESTRE Y SEA MAS FACIL DE ENTENDER POR EL USUARIO
        conversiones['Kelvin'] = resultado_k
        return conversiones
    elif unidad == 'f':
        resultado_c = (grados - 32) * (9/5)
        resultado_k = (grados - 32) * (9/5) + 273.15
        conversiones['Celsius'] = resultado_c
        conversiones['Kelvin'] = resultado_k
        return conversiones
    elif unidad == 'k':
        resultado_c = grados - 273.15
        resultado_f = ((grados - 273.15) * (9/5) + 32)
        conversiones['Celsius'] = resultado_c
        conversiones['Fahrenheit'] = resultado_f
        return conversiones
def validar_num(mensaje):
    '''Pide, valida y devuelve el número con el formato solicitado.'''
    while True:
        try:
            numero = float(input(mensaje))
            return numero
        except ValueError:
            print('Introduce un número, por favor')

def validar_letra(mensaje):
    '''Pide, normaliza, valida y devuelve la letra.'''
   
    while True:
        letra = input(mensaje).lower()
        if letra in ('c','f','k'):
            return letra
        else:
            print('Inserte la unidad correcta (C, K o F), por favor!!!')
        

grados = validar_num('Introduce el valor de la temperatura: ')
unidad = validar_letra('Indique la unidad de medida C(Celsius), F(Fahrenheit) o K(Kelvin): ')

resultado = conversor_temp(grados, unidad)
print('Conversiones obtenidas:\n')

for clave, valor in resultado.items():
    print(f'{clave}: {valor:.2f}')   
```

3. Contador de palabras de un texto.

```python
def contar_palabras(texto):
    '''Cuenta las apariciones de cada palabra y devuelve un diccionario con los valores.'''
    
    resultado = {}

    for palabra in texto.split():
        if palabra not in resultado:
            resultado[palabra] = 1
        else:
            resultado[palabra] +=1 
    return resultado

def calcular_long_media(texto):
    '''Calcula la longitud media de las palabras del texto.'''

    acumulador = 0
    contador = 0

    for palabra in texto.split():
        acumulador  += len(palabra)
        contador += 1

    media = acumulador / contador
    return media 

def limpiar_texto(texto):
    '''Normaliza el texto eliminando caracteres no alfabéticos y convirtiéndolo a minúsculas'''
    
    palabras_limpias = []

    for palabra in texto.lower().split():  
        palabra_limpia = ""  

        for letra in palabra:  
            if letra.isalpha():  # solo letras válidas (incluye á, é, í, ó, ú, ñ)
                palabra_limpia += letra

        if palabra_limpia:  # solo añadir si no quedó vacía
            palabras_limpias.append(palabra_limpia)

    return " ".join(palabras_limpias)

while True:
    texto = limpiar_texto(input('Introduce un texto: '))

    if texto:
        break
    else:
        print('Introduce un texto válido!!')

resultado = contar_palabras(texto)
print('La frecuencia de cada palabra sería: \n')

for clave, valor in resultado.items():
    print(f'{clave}: {valor}')

longitud = calcular_long_media(texto)
print(f'\nLa longitud media de las palabras del texto es {longitud:.2f} letras.')
```

4. Diccionario inverso (con posibilidad de duplicados)

```python
def crear_dicc_inverso(diccionario):
    '''Crea un diccionario inverso a partir de uno introducido e 
       informa en caso de haber claves duplicadas en el nuevo'''
    
    nuevo_diccionario = {}

    for clave, valor in diccionario.items():
        if valor not in nuevo_diccionario:
            nuevo_diccionario[valor] = clave
        else:
            if isinstance (nuevo_diccionario[valor], list):    #Evalua si existe una lista
                nuevo_diccionario[valor].append(clave)
            else:
                nuevo_diccionario[valor] = [nuevo_diccionario[valor],clave]
                 
    return nuevo_diccionario             
             
resultado = crear_dicc_inverso({'a':1, 'b':2, 'c':3, 'd':2})
print(resultado)
      
for clave, valor in resultado.items():
    if isinstance(valor,list):
        print(f'\nAdvertencia!!!: la clave inversa {clave} tiene asociados los valores {valor}.')
```

```python
archivo = 'tu me quieres blanca.txt'

with open(archivo, 'r', encoding='utf-8') as fichero:
    texto = fichero.read()

texto = limpiar_texto(texto)

resultado = contar_palabras(texto)

resultado_ordenado = sorted(resultado)    # Se crea una lista ordenada alfabéticamente

def organizar_dicc(resultado_ordenado, resultado):
    '''Crea un diccionario organizado por letra inicial a partir
       de una lista de palabras ordenadas y un diccionario de frecuencias.'''

    dicc_organizado = {}

    for palabra in resultado_ordenado:
        frecuencia = resultado[palabra]
        inicial = palabra[0]
        if inicial not in dicc_organizado:
            dicc_organizado[inicial] = {}
            dicc_organizado[inicial][palabra]= frecuencia
        else:
            dicc_organizado[inicial][palabra]= frecuencia

    return dicc_organizado

resultado_final = organizar_dicc(resultado_ordenado, resultado)

for clave, valor in resultado_final.items():
   print(f'{clave}: {valor}')
```

2. Conversión de tipos de datos.

```python
def convertir_datos(lista):
    '''Genera dos listas, una con valores convertidos a flotantes y la otra con los que no se pudieron convirtir.'''

    lista_flot = []
    lista_noflot = []

    for elemento in lista:
        if isinstance (elemento, (list, tuple)):    # isinstance(objeto,(tipo1, tipo2)) comprobar si es lista o tupla. 
            for dato in elemento:
                try:
                    modificado = float(dato)
                    lista_flot.append(modificado)
                except:
                    lista_noflot.append(dato)
        else:
            try:
                modificado = float(elemento)
                lista_flot.append(modificado)
            except:
                    lista_noflot.append(elemento)
           
    return lista_flot, lista_noflot


resultado = convertir_datos([ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']])
print(resultado)
```

1. Generador de Contraseñas

```python
import string
import numpy as np
import pyperclip

def generar_contraseña(longitud, mayuscula=True, minuscula=True, numeros=True, signos=False):
    '''Genera una contraseña aleatoria según los criterios
       indicados y devuelve el resultado como una cadena.'''
    
    lista_contraseña =[]

    letras_mayusculas = list(string.ascii_uppercase)        # Se convierte a lista las cadenas obtenidas 
    letras_minusculas = list(string.ascii_lowercase)
    digitos = list(string.digits)
    caract_signos = list(string.punctuation)

    if mayuscula:     # Añadir una mayúscula (si procede)
        letra_mayuscula = np.random.choice(letras_mayusculas)
        lista_contraseña.append(letra_mayuscula)

    if minuscula:     # Añadir una minúscula 
        letra_minuscula = np.random.choice(letras_minusculas)
        lista_contraseña.append(letra_minuscula)

    if numeros:       # Añadir un número
        digito = np.random.choice(digitos)
        lista_contraseña.append(digito)

    if signos:        # Añadir un signo
        signo = np.random.choice(caract_signos)
        lista_contraseña.append(signo) 

    caracteres_restantes = longitud - len(lista_contraseña)     

    caracteres_disponibles = []          # Se crea una lista de caracteres disponibles

    if mayuscula:
        caracteres_disponibles +=  letras_mayusculas
    if minuscula:
        caracteres_disponibles += letras_minusculas

    if numeros:
        caracteres_disponibles += digitos

    if signos:
        caracteres_disponibles += caract_signos

    for caracter in range(caracteres_restantes):              #Completar hasta la longitud deseada
        caracter = np.random.choice(caracteres_disponibles)
        lista_contraseña.append(caracter)

    np.random.shuffle(lista_contraseña)          # Mezclar todos los caracteres
    
    contraseña = "".join(lista_contraseña)      # Unirlos en un único string
     
    pyperclip.copy(contraseña)          # Copia automáticamente la contraseña al portapapeles

    return contraseña      #Devolver la contraseña

print(generar_contraseña(10, True, True, True, True))
```

2. Procesamiento de datos simple

```python
def procesar_partidos(archivo):
    '''Procesa un histórico de partidos de fútbol y devuelve:
    Goles marcados por cada equipo, equipo más goleador, equipo más goleado.
    y la clasificación por puntos.'''

    goles_marcados = {}
    goles_contra = {}
    total_puntos = {}

    with open(archivo, 'r', encoding='utf-8') as fichero:

        for linea in fichero:

            linea = linea.strip()
            partido = linea.split("\t")

            equipo_local, resultado, equipo_visitante = partido

            resultado = resultado.split("-")
            goles_locales, goles_visitante = resultado

            goles_locales = int(goles_locales)
            goles_visitante = int(goles_visitante)

            # GOLES MARCADOS

            if equipo_local not in goles_marcados:
                goles_marcados[equipo_local] = 0

            goles_marcados[equipo_local] += goles_locales

            if equipo_visitante not in goles_marcados:
                goles_marcados[equipo_visitante] = 0

            goles_marcados[equipo_visitante] += goles_visitante

            # GOLES EN CONTRA

            if equipo_local not in goles_contra:
                goles_contra[equipo_local] = 0

            goles_contra[equipo_local] += goles_visitante

            if equipo_visitante not in goles_contra:
                goles_contra[equipo_visitante] = 0

            goles_contra[equipo_visitante] += goles_locales

            # PUNTOS

            if equipo_local not in total_puntos:
                total_puntos[equipo_local] = 0

            if equipo_visitante not in total_puntos:
                total_puntos[equipo_visitante] = 0

            if goles_locales > goles_visitante:
                total_puntos[equipo_local] += 3

            elif goles_locales < goles_visitante:
                total_puntos[equipo_visitante] += 3

            else:
                total_puntos[equipo_local] += 1
                total_puntos[equipo_visitante] += 1

    # RESULTADOS FINALES

    equipo_mas_goleador = max(goles_marcados, key=goles_marcados.get)
    equipo_mas_goleado = max(goles_contra, key=goles_contra.get)

    clasificacion = sorted(total_puntos.items(),key=lambda equipo: equipo[1], reverse=True)

    return (goles_marcados, equipo_mas_goleador, equipo_mas_goleado, clasificacion)

archivo = "historic partits.txt"         # PROGRAMA PRINCIPAL

goles, mas_goleador, mas_goleado, clasificacion = procesar_partidos(archivo)

print("GOLES MARCADOS")
for equipo, total in goles.items():
    print(f"{equipo}: {total}")

print("\nEquipo más goleador:")
print(mas_goleador)

print("\nEquipo más goleado:")
print(mas_goleado)

print("\nCLASIFICACIÓN")
for posicion, (equipo, puntos) in enumerate(clasificacion, start=1):
    print(f"{posicion}. {equipo}: {puntos} puntos")
```

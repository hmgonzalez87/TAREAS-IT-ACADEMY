Se cargan todos los archivos CSV contenidos en la carpeta dataset en un diccionario de DataFrames. Cada clave corresponde al nombre del archivo (sin la extensión .csv) y el valor es el DataFrame asociado. Esta estructura facilita la gestión de múltiples tablas durante el análisis.

```python
ruta = Path("..") / "Análisis de Operaciones Logísticas" / "Data" / "dataset"

dataframes = {}

for archivo in ruta.glob("*.csv"):
    dataframes[archivo.stem] = pd.read_csv(archivo)
```

```python
list(dataframes.keys())           # Se genera lista con los nombres de los archivos que tenemos que serían las claves del diccionario.
```

Para facilitar la lectura del código, cada DataFrame se asigna a una variable con el mismo nombre que su archivo de origen.

```python
customers = dataframes['customers']
delivery_events = dataframes['delivery_events']
drivers = dataframes['drivers']
driver_monthly_metrics = dataframes['driver_monthly_metrics']
facilities = dataframes['facilities']
fuel_purchases = dataframes['fuel_purchases']
loads = dataframes['loads']
maintenance_records = dataframes['maintenance_records']
routes = dataframes['routes']
safety_incidents = dataframes['safety_incidents']
trailers = dataframes['trailers']
trips = dataframes['trips']
trucks = dataframes['trucks']
truck_utilization_metrics = dataframes['truck_utilization_metrics']
```
